```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#load_dataset in csv format
df = pd.read_csv("/content/student_data.csv")
df.head(50)
```

```python
#structure of load_data
df.info()
```

```python
# removing unused columns
df = df.drop(columns=["Fedu","goout"],axis = 1, errors = "ignore")
```

```python
#add new coulmns
df["studytime"]= range(len(df))
df['qualification']=range(len(df))
df["studytime"]=np.random.randint(1,10,size=len(df))

df.head(10)
```

```python
#Changes and update columns
df["attendence"]=df["attendence"].replace('T','P')

df["marks"]= np.random.randint(0,100,size = len(df))

df = df.rename(columns={"qualification":"grades"})

df["grades"]=np.where(df['marks']<40,0,1)

df.head(20)
```

```python
# shapes of load_dataset
print(df.shape)
```

```python
#handling missing values
x = df.isnull()
print(x)
```

```python
#check for duplicate
df.duplicated()
```

```python
#removes duupicate rows
df = df.drop_duplicates()
df.info()
```

```python
#finding categorical and numerical data
cat_col = [ data for data in df.columns if df[data].dtype == object]
num_col = [ data for data in df.columns if df[data].dtype != object]
print(cat_col)
print(num_col)
```

```python
# check all statistical parameters
df.describe()
```

```python
# finding Average values of marks
print("average marks = ", df['marks'].mean())
```

```python
# ---------------------------------------------------------------visualizations-----------------------------------------------------------------
# figure initialization
plt.figure(figsize=(15,3))

#histograph of grades
plt.subplot(1,3,1)
plt.hist(df["grades"],bins=10,color='skyblue', edgecolor='black')
plt.title('Distribution of Grades')
plt.xlabel('Numeric Grade Value')
plt.ylabel('Number of Students')
plt.show()

```

```python
# scatter of points of studytime and grades
plt.subplot(1, 3, 2)
plt.scatter(df["studytime"], df["grades"], color='purple', alpha=0.6)
plt.title('Study Time vs Grades')
plt.xlabel('Study Time (Hours)')
plt.ylabel('Grades')
```

```python
#finding male and female avg marks
male_avg = df[df['sex'] == 'M']['marks'].mean()
female_avg = df[df['sex'] == 'F']['marks'].mean()
print("male Average marks =",male_avg)
print("male Average marks =",female_avg)
```

```python
# Bar Chart: Male vs Female Average Score
plt.subplot(1, 3, 3)
plt.bar(["M","F"], [male_avg, female_avg], color=['blue', 'pink'])
plt.title('Avg Score: Male vs Female',size=10)
plt.ylabel('Average Grade')
plt.ylim(0, 100) # Set Y-axis limit for better comparison
```
